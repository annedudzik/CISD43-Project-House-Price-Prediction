#### The objective of this notebook is to load the house prediction dataset into MongoDB and make queries.

#### Steps:

#### Create a Database and a Collection

#### I created a new database and collection in MongoDB with the house price prediction data (see pdf images file)

#### Install and import necessary tools

```python
# already installed
# pip install pymongo
from pymongo import MongoClient
import pandas as pd
```

#### Connect to MongoDB

```python
client = MongoClient('mongodb://localhost:27017/')
```

#### Select the database House_Price_Prediction_db

```python
db = client['House_Price_Prediction_db']
```

#### Select the collection House_data

```python
collection = db['House_data']
```

#### Retrieve the first 4 records

```python
# Retrieve the first four records (the first four documents)
result = collection.find().limit(4)

# print the records
for record in result:
    print(record)
```

#### Query: Find the document with id 1175000570

```python
# Find one document: 
# define the query filter
query = {'id':1175000570}

# Retrieve the document with id 1175000570
result = collection.find_one(query)
print(result)
```

#### Create a dataframe using all the records from the collection

```python
# Retrieve all records
result = collection.find({})

# Convert the result to a pandas Dataframe
df = pd.DataFrame(list(result))
print(df)
```
